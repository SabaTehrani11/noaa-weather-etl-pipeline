```python
import pandas as pd
import os

# Define folder path
path = "/Users/sabatehrani/Desktop/Historical_Weather_Data/archive (3)/NOAA Weather cities 2015-2020/"


def extract_data(path):

    files = [
        "Weather all cities 2015.csv",
        "Weather all cities 2016.csv",
        "Weather all cities 2017.csv",
        "Weather all cities 2018.csv",
        "Weather all cities 2019.csv",
        "Weather all cities 2020.csv"
    ]


    # Step 1: Check if files exist
    for file in files:

        if not os.path.exists(path + file):
            print(file + " does not exist")
            continue

        else:
            print(file + " exists")


    # Step 2: Read CSV files and check if they are empty
    has_empty_file = False

    df_2015 = pd.read_csv(path + "Weather all cities 2015.csv")
    if df_2015.empty:
        print("df_2015 is empty")
        has_empty_file = True

    df_2016 = pd.read_csv(path + "Weather all cities 2016.csv")
    if df_2016.empty:
        print("df_2016 is empty")
        has_empty_file = True

    df_2017 = pd.read_csv(path + "Weather all cities 2017.csv")
    if df_2017.empty:
        print("df_2017 is empty")
        has_empty_file = True

    df_2018 = pd.read_csv(path + "Weather all cities 2018.csv")
    if df_2018.empty:
        print("df_2018 is empty")
        has_empty_file = True

    df_2019 = pd.read_csv(path + "Weather all cities 2019.csv")
    if df_2019.empty:
        print("df_2019 is empty")
        has_empty_file = True

    df_2020 = pd.read_csv(path + "Weather all cities 2020.csv")
    if df_2020.empty:
        print("df_2020 is empty")
        has_empty_file = True


    # Step 3: Report if one or more files are empty
    if has_empty_file == True:
        print("One or more files are empty")


    # Step 4: Combine non-empty files
    valid_dataframes = [
        df for df in
        [df_2015, df_2016, df_2017, df_2018, df_2019, df_2020]
        if not df.empty
    ]

    df = pd.concat(
        valid_dataframes,
        ignore_index=True
    )


    # Step 5: Check required columns
    has_required_columns = True

    if "Date" not in df.columns:
        print("Date column is missing")
        has_required_columns = False

    if "ID" not in df.columns:
        print("ID column is missing")
        has_required_columns = False


    # Step 6: Return extracted data
    if has_required_columns == True:
        return df
```

### Step 2: Clean Data

```python
def clean_data(df):

    # Make a copy
    df = df.copy()

    # Check duplicates before removing
    print("Duplicated before:", df.duplicated().sum())

    # Remove duplicates
    df = df.drop_duplicates()

    # Check duplicates after removing
    print("Duplicated after:", df.duplicated().sum())

    # Check missing values before removing
    print("Missing values before:")
    print(df.isnull().sum())

    # Remove null values from important columns
    df = df.dropna(subset=["Date", "ID"])

    # Check missing values after removing
    print("Missing values after:")
    print(df.isnull().sum())

    return df
```

### Step 3: Transform and Standardize Data

```python
def transform_data(clean_df):

    # Make a copy
    transformed_df = clean_df.copy()

    # Standardize Date
    transformed_df["Date"] = pd.to_datetime(
        transformed_df["Date"],
        format="%Y%m%d",
        errors="coerce"
    )

    # Standardize Temperature
    transformed_df["TAVG"] = transformed_df["TAVG"] / 10
    transformed_df["TMAX"] = transformed_df["TMAX"] / 10
    transformed_df["TMIN"] = transformed_df["TMIN"] / 10

    return transformed_df
```

### Step 4: Data Quality Checks

```python
def check_data_quality(transformed_df):

    # Check for duplicate rows
    duplicate_count = transformed_df.duplicated().sum()

    # Check for missing values in important columns
    missing_values = transformed_df[["Date", "ID"]].isnull().sum().sum()

    # Stop pipeline if data quality check fails
    if duplicate_count > 0 or missing_values > 0:
        raise ValueError("Data quality check failed")
    else:
        print("Data quality check passed!")
```

### Step 5: Load Data

```python
from sqlalchemy import create_engine
```

```python
def load_data(transformed_df):

    # Create connection to PostgreSQL
    engine = create_engine(
    "postgresql+psycopg2://postgres:sabatehrani22647741@localhost:5432/weather_etl_db"
)
    # Load data into PostgreSQL
    transformed_df.to_sql(
    "weather_data",
    engine,
    if_exists = "replace",
    index = False,
    chunksize = 10000
    )

    print("Data loaded successfully!")
```

### Step 6: Build Complete ETL Pipeline¶

```python
def run_pipeline(path):

    # Extract
    df = extract_data(path)

    # Clean
    clean_df = clean_data(df)

    # Transform
    transformed_df = transform_data(clean_df)

    # Data Quality
    check_data_quality(transformed_df)

    # Load
    load_data(transformed_df)

    print("ETL pipeline completed successfully!")
```

```python
run_pipeline(path)
```

### Step 7: Add Logging and Error Handling

```python
import logging
import os
```

```python
# Create logs folder
os.makedirs("logs", exist_ok=True)

# Configure logging
logging.basicConfig(
    filename="logs/pipeline.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)
```

```python
def run_pipeline(path):

    try:
        # Start pipeline
        logging.info("ETL pipeline started")

        # Extract
        df = extract_data(path)
        logging.info(f"{len(df)} records extracted successfully")

        # Clean
        clean_df = clean_data(df)
        logging.info(f"{len(clean_df)} records cleaned successfully")

        # Transform
        transformed_df = transform_data(clean_df)
        logging.info(f"{len(transformed_df)} records transformed successfully")

        # Data Quality
        check_data_quality(transformed_df)
        logging.info("Data quality check passed")

        # Load
        load_data(transformed_df)
        logging.info("Data successfully loaded into PostgreSQL")

        # Complete pipeline
        logging.info("ETL pipeline completed successfully!")
        print("ETL pipeline completed successfully!")

    except Exception as e:
        logging.error(f"Pipeline failed: {e}")
        print("Pipeline failed:", e)
```

```python
run_pipeline(path)
```
